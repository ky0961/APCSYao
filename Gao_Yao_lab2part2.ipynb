```python
# TODO: Feature extraction function here.

# SAMPLE SCAFFOLDING CODE FOR EXTRACTION PIPELINE FUNCTION (Modify as needed for
# your specific environment)
def extract(model, data_loader, conv_2 = 5, conv_5 = 12, device=torch.device("cuda")):
  pool2_features, pool5_features, labels = [], [], []
  with torch.no_grad():
    for images, y in data_loader:
      images = images.to(device)
      x = images
      for i, layer in enumerate(model.features):
        x = layer(x)
        if i == conv_2:
          pool2_features.append(torch.flatten(x, start_dim=1).cpu())
        if i == conv_5: # After the 12th layer (ReLU)
          pool5_features.append(torch.flatten(x, start_dim=1).cpu())
      labels.append(y)
    pool2_features = torch.cat(pool2_features, dim=0)
  pool5_features = torch.cat(pool5_features, dim=0)
  labels = torch.cat(labels, dim=0)
  return pool2_features, pool5_features, labels
```
